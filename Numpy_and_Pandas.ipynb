Data Analysis Using NumPy and Pandas

```python
import numpy as np
```

## 1. NumPy Array Operations

### Creating a 1D NumPy Array

A 1D NumPy array named `temperatures_w1` is created to store the average daily temperatures for Week 1.

```python
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print(temperatures_w1)
```

### Inspecting the Array

The shape, data type, and number of elements of the Week 1 temperature array are inspected using NumPy properties.

```python
print("Shape:", temperatures_w1.shape)
print("Data Type:", temperatures_w1.dtype)
print("Number of Elements:", temperatures_w1.size)
```

### Converting Celsius to Fahrenheit

The Celsius temperatures are converted to Fahrenheit using the formula:

Fahrenheit = (Celsius × 9/5) + 32

```python
temperatures_f = (temperatures_w1 * 9/5) + 32

print("Temperatures in Fahrenheit:")
print(temperatures_f)
```

### Finding Maximum, Minimum, and Mean Temperatures

NumPy functions are used to calculate the maximum, minimum, and average temperature for Week 1.

```python
print("Maximum Temperature:", np.max(temperatures_w1))
print("Minimum Temperature:", np.min(temperatures_w1))
print("Mean Temperature:", np.mean(temperatures_w1))
```

### Array Slicing and Indexing

Array slicing is used to extract the first three days, weekend temperatures, and middle three days from the Week 1 temperature array.

```python
# First three days
print("First three days:", temperatures_w1[:3])

# Weekend - last two days
print("Weekend:", temperatures_w1[-2:])

# Middle three days
print("Middle three days:", temperatures_w1[2:5])
```

## 2. Creating a 2D NumPy Array

A 2D NumPy array named `temperatures` is created. Each row represents the daily temperatures for one week.

```python
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])

print(temperatures)
```

### Inspecting the 2D Array

The shape, data type, and total number of elements in the 2D temperature array are displayed.

```python
print("Shape:", temperatures.shape)
print("Data Type:", temperatures.dtype)
print("Total Number of Elements:", temperatures.size)
```

### Extracting Weekly and Weekend Temperatures

The temperatures for each week and the weekend temperatures for both weeks are extracted using NumPy indexing and slicing.

```python
print("Week 1:", temperatures[0])
print("Week 2:", temperatures[1])

print("Week 1 Weekend:", temperatures[0, -2:])
print("Week 2 Weekend:", temperatures[1, -2:])
```

## 3. Pandas Series

Pandas Series is used to store and manipulate the student marks with custom index labels.

```python
import pandas as pd
```

### Creating a Pandas Series

A Pandas Series named `marks` is created with student marks and custom index labels representing their ranks.

```python
marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

print(marks)
```

### Indexing the Pandas Series

The `iloc` accessor is used to access the mark of the 1st rank student using its integer position

```python
print("Mark of 1st Rank:", marks.iloc[0])
```

### Retrieving the Top 3 Ranks

The `loc` accessor is used to retrieve the marks of the top three ranks using their index labels.

```python
top_three = marks.loc[['Rank1', 'Rank2', 'Rank3']]

print(top_three)
```

### Retrieving the 3rd-Ranked Student

The `iloc` accessor is used to retrieve the mark of the student at the third integer position.

```python
print("Mark of 3rd Rank:", marks.iloc[2])
```

### Boolean Filtering

A boolean condition is applied to the Series to retrieve students whose marks are greater than 90.

```python
high_marks = marks[marks > 90]

print("Students with marks greater than 90:")
print(high_marks)
```

### Modifying a Series Value

The mark of the student with the 1st rank is modified from 95 to 100.

```python
marks.loc['Rank1'] = 100

print(marks)
```

### Removing the Last Rank

The entry corresponding to Rank5 is removed from the Series using the `drop()` method.

```python
marks = marks.drop('Rank5')

print(marks)
```

### Calculating CGPA

The CGPA is calculated by dividing each student's mark by 10.

```python
cgpa = marks / 10

print("CGPA:")
print(cgpa)
```

## 4. Pandas DataFrame

A Pandas DataFrame is created to store and analyze transaction data containing transaction IDs, product categories, regions, and amounts.

```python
transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],

    'ProductCategory': [
        'Electronics',
        'Clothing',
        'Electronics',
        'Furniture',
        'Clothing',
        'Electronics',
        'Furniture',
        'Clothing',
        'Furniture',
        'Electronics'
    ],

    'Region': [
        'North',
        'South',
        'North',
        'East',
        'West',
        'North',
        'East',
        'West',
        'South',
        'North'
    ],

    'Amount': [
        200,
        150,
        300,
        450,
        200,
        250,
        300,
        180,
        350,
        400
    ]
})

print(transactions)
```

### Data Exploration

The first five rows of the transactions DataFrame are displayed using the `head()` method.

```python
print(transactions.head())
```

### Displaying the Last Five Rows

The last five rows of the DataFrame are displayed using the `tail()` method.

```python
print(transactions.tail())
```

### Inspecting the DataFrame

The shape, column names, and data types of the transactions DataFrame are inspected.

```python
print("Shape:", transactions.shape)

print("\nColumn Names:")
print(transactions.columns)

print("\nData Types:")
print(transactions.dtypes)
```

### Basic Information

The `info()` method is used to display basic information about the DataFrame, including the number of entries, columns, and data types.

```python
transactions.info()
```

### Selecting Specific Columns

Only the `ProductCategory` and `Amount` columns are selected from the DataFrame.

```python
print(transactions[['ProductCategory', 'Amount']])
```

### Retrieving the Last Three Columns

The last three columns of the DataFrame are retrieved using the `iloc` accessor.

```python
print(transactions.iloc[:, -3:])
```

### Filtering Transactions

The DataFrame is filtered to find transactions where the region is North and the amount is greater than 200.

```python
filtered_transactions = transactions[
    (transactions['Region'] == 'North') &
    (transactions['Amount'] > 200)
]

print(filtered_transactions)
```

### Product Category Value Counts

The `value_counts()` method is used to count how many transactions belong to each product category.

```python
category_counts = transactions['ProductCategory'].value_counts()

print(category_counts)
```

### Finding Unique Regions

The `unique()` method is used to identify all the different regions present in the transactions DataFrame.

```python
unique_regions = transactions['Region'].unique()

print("Unique Regions:")
print(unique_regions)
```

### Grouping Transactions by Region

The transactions are grouped by region and the mean transaction amount is calculated for each region.

```python
mean_amount_by_region = transactions.groupby('Region')['Amount'].mean()

print(mean_amount_by_region)
```

## 5. DataFrame Manipulation

### Modifying a Transaction Amount

The amount of the transaction with TransactionID 102 is changed from 150 to 165.

```python
transactions.loc[
    transactions['TransactionID'] == 102,
    'Amount'
] = 165

print(transactions)
```

### Adding a Discount Column

A new column named `Discount` is added to the DataFrame. The discount is calculated as 10% of the transaction amount.

```python
transactions['Discount'] = transactions['Amount'] * 0.10

print(transactions)
```

### Removing a Transaction

The row corresponding to TransactionID 109 is removed from the DataFrame.

```python
transactions = transactions[
    transactions['TransactionID'] != 109
]

print(transactions)
```

### Removing the Discount Column

The `Discount` column is deleted from the DataFrame using the `drop()` method.

```python
transactions = transactions.drop(columns=['Discount'])

print(transactions)
```

## 6. Final DataFrame

The final transactions DataFrame after all modifications is displayed below.

```python
print(transactions)
```
